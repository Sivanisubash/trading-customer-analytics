feature engineering

```python
trades["trade_date"] = trades["trade_timestamp"].dt.date
trades["trade_month"] = (
    trades["trade_timestamp"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
import numpy as np

trades["revenue_usd"] = np.where(
    trades["pnl_usd"] < 0,
    abs(trades["pnl_usd"]),
    0
)
```

```python
trade_summary = (
    trades.groupby("customer_id")
    .agg(
        trade_count=("trade_id", "count"),
        trading_volume_usd=("stake_usd", "sum"),
        total_customer_pnl_usd=("pnl_usd", "sum"),
        revenue_usd=("revenue_usd", "sum"),
        average_trade_value_usd=("stake_usd", "mean"),
        last_trade_date=("trade_timestamp", "max"),
        active_trading_days=("trade_date", "nunique")
    )
    .reset_index()
)

trade_summary.head()
```

```python
payments["payment_date"] = payments["payment_timestamp"].dt.date
payments["payment_month"] = (
    payments["payment_timestamp"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
deposit_summary = (
    payments[
        (payments["payment_type"] == "Deposit") &
        (payments["payment_status"] == "Completed")
    ]
    .groupby("customer_id")
    .agg(
        total_deposit_usd=("amount_usd", "sum"),
        average_deposit_usd=("amount_usd", "mean"),
        deposit_count=("payment_id", "count")
    )
    .reset_index()
)
```

```python
withdrawal_summary = (
    payments[
        (payments["payment_type"] == "Withdrawal") &
        (payments["payment_status"] == "Completed")
    ]
    .groupby("customer_id")
    .agg(
        total_withdrawal_usd=("amount_usd", "sum"),
        average_withdrawal_usd=("amount_usd", "mean"),
        withdrawal_count=("payment_id", "count")
    )
    .reset_index()
)
```

```python
logins["login_date"] = logins["login_timestamp"].dt.date
logins["login_month"] = (
    logins["login_timestamp"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
login_summary = (
    logins.groupby("customer_id")
    .agg(
        login_count=("login_id", "count"),
        successful_logins=("login_success", "sum"),
        active_login_days=("login_date", "nunique"),
        average_session_minutes=("session_minutes", "mean"),
        last_login_date=("login_timestamp", "max")
    )
    .reset_index()
)

login_summary.head()
```

```python
customer_analytics = (
    customers
    .merge(trade_summary, on="customer_id", how="left")
    .merge(deposit_summary, on="customer_id", how="left")
    .merge(withdrawal_summary, on="customer_id", how="left")
    .merge(login_summary, on="customer_id", how="left")
    .merge(campaigns, on="campaign_id", how="left")
)
```

```python
numeric_columns = [
    "trade_count",
    "trading_volume_usd",
    "total_customer_pnl_usd",
    "revenue_usd",
    "average_trade_value_usd",
    "active_trading_days",
    "total_deposit_usd",
    "average_deposit_usd",
    "deposit_count",
    "total_withdrawal_usd",
    "average_withdrawal_usd",
    "withdrawal_count",
    "login_count",
    "successful_logins",
    "active_login_days",
    "average_session_minutes"
]

customer_analytics[numeric_columns] = (
    customer_analytics[numeric_columns].fillna(0)
)
```

```python
print(customer_analytics.shape)
customer_analytics.head()
```

```python
analysis_date = pd.Timestamp("2026-07-01")
```

```python
customer_analytics["days_since_last_trade"] = (
    analysis_date - customer_analytics["last_trade_date"]
).dt.days

customer_analytics["days_since_last_login"] = (
    analysis_date - customer_analytics["last_login_date"]
).dt.days
```

```python
customer_analytics[
    [
        "customer_id",
        "last_trade_date",
        "days_since_last_trade",
        "last_login_date",
        "days_since_last_login"
    ]
].head()
```

```python
def assign_activity_status(days_since_last_trade):
    if pd.isna(days_since_last_trade):
        return "Never Traded"
    elif days_since_last_trade <= 30:
        return "Active"
    elif days_since_last_trade <= 60:
        return "At Risk"
    elif days_since_last_trade <= 90:
        return "Dormant"
    else:
        return "Churned"

customer_analytics["activity_status"] = (
    customer_analytics["days_since_last_trade"]
    .apply(assign_activity_status)
)
```

```python
customer_analytics["activity_status"].value_counts()
```

```python
def assign_customer_segment(row):
    if row["trade_count"] == 0:
        return "Non-Trader"

    elif (
        row["trading_volume_usd"] >= 10000
        and row["trade_count"] >= 50
    ):
        return "VIP Trader"

    elif row["trading_volume_usd"] >= 5000:
        return "High-Value Trader"

    elif row["trade_count"] >= 20:
        return "Frequent Trader"

    else:
        return "Occasional Trader"

customer_analytics["customer_segment"] = (
    customer_analytics.apply(
        assign_customer_segment,
        axis=1
    )
)
```

```python
customer_analytics["customer_segment"].value_counts()
```

```python
customer_analytics["account_age_days"] = (
    analysis_date - customer_analytics["signup_date"]
).dt.days

customer_analytics["account_age_months"] = (
    customer_analytics["account_age_days"] / 30.44
).clip(lower=1)
```

```python
customer_analytics["monthly_revenue_usd"] = (
    customer_analytics["revenue_usd"]
    / customer_analytics["account_age_months"]
)

customer_analytics["estimated_clv_usd"] = (
    customer_analytics["monthly_revenue_usd"] * 24
)
```

```python
customer_analytics["net_deposit_usd"] = (
    customer_analytics["total_deposit_usd"]
    - customer_analytics["total_withdrawal_usd"]
)
```

```python
customer_analytics["churn_risk_score"] = (
    np.where(
        customer_analytics["days_since_last_trade"] > 60,
        35,
        0
    )
    +
    np.where(
        customer_analytics["days_since_last_login"] > 30,
        25,
        0
    )
    +
    np.where(
        customer_analytics["trade_count"] < 5,
        15,
        0
    )
    +
    np.where(
        customer_analytics["total_deposit_usd"] == 0,
        15,
        0
    )
    +
    np.where(
        customer_analytics["active_login_days"] < 5,
        10,
        0
    )
)
```

```python
customer_analytics["churn_risk_level"] = pd.cut(
    customer_analytics["churn_risk_score"],
    bins=[-1, 29, 59, 100],
    labels=["Low", "Medium", "High"]
)
```

```python
customer_analytics["churn_risk_level"].value_counts()
```

```python
customer_analytics["profitable_trader_flag"] = np.where(
    customer_analytics["total_customer_pnl_usd"] > 0,
    "Profitable",
    "Unprofitable"
)
```

```python
customer_analytics["high_value_customer_flag"] = np.where(
    (
        customer_analytics["trading_volume_usd"] >= 5000
    )
    |
    (
        customer_analytics["estimated_clv_usd"] >= 1000
    ),
    "High Value",
    "Standard Value"
)
```
